```python
import pandas as pd

import nltk

import nlpaug.augmenter.word as naw
```

```python
nltk.download('punkt')

nltk.download('wordnet')

nltk.download('omw-1.4')

nltk.download('averaged_perceptron_tagger')

nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk import pos_tag

from nltk.tokenize import word_tokenize

text = "I feel hopeless and lonely"

tokens = word_tokenize(text)

print(pos_tag(tokens))
```

```python
df=pd.read_csv("../dataset/processed_dataset.csv")
```

```python
df.shape
```

```python
df.head()
```

```python
aug = naw.SynonymAug(
    aug_src='wordnet',
    aug_p=0.05
)
```

```python

for i in range(5):
    print(aug.augment("I feel hopeless and lonely"))
```

```python
sample_df = df.sample(15000, random_state=42)
```

```python
sample_df['synthetic_text'] = sample_df['text'].apply(
    lambda x: aug.augment(x)[0]
)
```

```python
sample_df[['text', 'synthetic_text']].head()
```

```python
synthetic_df = pd.DataFrame({
    'text': sample_df['synthetic_text'],
    'class': sample_df['class']
})
```

```python
augmented_df = pd.concat(
    [df, synthetic_df],
    ignore_index=True
)
```

```python
print(augmented_df.shape)
```

```python
augmented_df.to_csv("../dataset/augmented_dataset.csv",index=False)
```
