```python
from tensorflow.keras.models import load_model

model = load_model("../trained_models/lstm_model.h5")
```

```python
import pickle

with open("../trained_models/tokenizer.pkl", "rb") as f:
    tokenizer = pickle.load(f)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
def predict_suicide(text):

    sequence = tokenizer.texts_to_sequences([text])

    padded = pad_sequences(sequence, maxlen=100)

    prediction = model.predict(padded)[0][0]

    if prediction > 0.5:
        print("Prediction: Suicide")
    else:
        print("Prediction: Non-Suicide")

    print("Probability:", prediction)
```

```python
predict_suicide("Life is beautiful")

predict_suicide("Nobody understands me")

predict_suicide("I want to disappear")
```

```python
predict_suicide("Everything feels pointless, I can’t go on.”")
```
